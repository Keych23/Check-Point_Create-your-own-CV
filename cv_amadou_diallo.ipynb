## 🎨 2. Rendu HTML du CV

```python
from IPython.display import HTML, display

def render_cv():
    """Génère et affiche le CV complet en HTML dans Colab."""

    # ── CSS ──────────────────────────────────────────────────────────
    css = """
    @import url('https://fonts.googleapis.com/css2?family=DM+Serif+Display:ital@0;1&family=DM+Mono:wght@400;500&family=Syne:wght@400;600;700&display=swap');
    .cv-wrap * { box-sizing: border-box; margin: 0; padding: 0; }
    .cv-wrap {
      font-family: 'Syne', sans-serif;
      background: #f8f7f4;
      padding: 24px;
      -webkit-font-smoothing: antialiased;
    }
    .cv-page {
      max-width: 860px; margin: 0 auto;
      background: #fff;
      box-shadow: 0 4px 40px rgba(0,0,0,0.08);
    }
    /* Header */
    .cv-header {
      display: grid; grid-template-columns: 1fr auto; gap: 2rem;
      align-items: end; padding: 3rem 3rem 2rem;
      border-bottom: 1.5px solid #0e0e10;
      position: relative;
    }
    .cv-header::before {
      content: ''; position: absolute; top:0; left:0; right:0;
      height: 5px; background: #1a3cff;
    }
    .cv-name {
      font-family: 'DM Serif Display', serif;
      font-size: 3rem; line-height: 1.05;
      letter-spacing: -0.02em; color: #0e0e10;
    }
    .cv-name em { font-style: italic; color: #1a3cff; }
    .cv-role {
      font-family: 'DM Mono', monospace; font-size: 0.82rem;
      color: #1a3cff; letter-spacing: 0.08em;
      text-transform: uppercase; margin-top: 0.6rem;
    }
    .cv-summary {
      font-size: 0.9rem; color: #5a5a6a;
      line-height: 1.7; max-width: 420px; margin-top: 0.6rem;
    }
    .cv-badge {
      display: inline-block; background: #0e0e10; color: #fff;
      font-family: 'DM Mono', monospace; font-size: 0.68rem;
      padding: 4px 10px; border-radius: 2px;
      letter-spacing: 0.06em; margin-top: 0.5rem;
    }
    .cv-contact {
      text-align: right; font-family: 'DM Mono', monospace;
      font-size: 0.75rem; color: #5a5a6a; line-height: 2;
    }
    .cv-contact a { color: #1a3cff; text-decoration: none; }
    /* Body grid */
    .cv-body { display: grid; grid-template-columns: 1fr 250px; }
    .cv-main { padding: 2.5rem 2.5rem 3rem; border-right: 1px solid #e2e0da; }
    .cv-aside { padding: 2.5rem 1.75rem 3rem; background: #fafaf8; }
    /* Section */
    .cv-section { margin-bottom: 2.25rem; }
    .cv-section:last-child { margin-bottom: 0; }
    .cv-section-title {
      font-family: 'DM Mono', monospace; font-size: 0.68rem;
      font-weight: 500; letter-spacing: 0.12em;
      text-transform: uppercase; color: #1a3cff;
      margin-bottom: 1.1rem; display: flex;
      align-items: center; gap: 8px;
    }
    .cv-section-title::after {
      content: ''; flex: 1; height: 1px; background: #e2e0da;
    }
    /* Job */
    .cv-job { margin-bottom: 1.6rem; padding-bottom: 1.6rem; border-bottom: 1px solid #e2e0da; }
    .cv-job:last-child { border-bottom: none; margin-bottom: 0; padding-bottom: 0; }
    .cv-job-head { display: flex; justify-content: space-between; align-items: baseline; gap: 1rem; }
    .cv-job-title {
      font-family: 'DM Serif Display', serif;
      font-size: 1.1rem; color: #0e0e10;
    }
    .cv-job-period {
      font-family: 'DM Mono', monospace; font-size: 0.7rem;
      color: #9898a8; white-space: nowrap;
    }
    .cv-job-company {
      font-size: 0.8rem; color: #1a3cff; font-weight: 600;
      margin-top: 2px; margin-bottom: 0.5rem;
    }
    .cv-job-desc { font-size: 0.85rem; color: #5a5a6a; line-height: 1.65; }
    .cv-achievements { margin-top: 0.5rem; padding-left: 0; list-style: none; }
    .cv-achievements li {
      font-size: 0.82rem; color: #0e0e10;
      padding: 3px 0 3px 1rem; position: relative; line-height: 1.5;
    }
    .cv-achievements li::before {
      content: '→'; position: absolute; left: 0;
      color: #1a3cff; font-size: 0.72rem;
    }
    /* Stats */
    .cv-stats { display: grid; grid-template-columns: 1fr 1fr; gap: 8px; margin-bottom: 1.5rem; }
    .cv-stat {
      background: #e8edff; border-radius: 4px;
      padding: 10px 12px; text-align: center;
    }
    .cv-stat-num {
      font-family: 'DM Serif Display', serif;
      font-size: 1.45rem; color: #1a3cff; display: block;
    }
    .cv-stat-lbl {
      font-family: 'DM Mono', monospace; font-size: 0.62rem;
      text-transform: uppercase; letter-spacing: 0.08em;
      color: #5a5a6a; margin-top: 2px;
    }
    /* Tags */
    .cv-tag-group { margin-bottom: 1.1rem; }
    .cv-tag-group-lbl {
      font-family: 'DM Mono', monospace; font-size: 0.66rem;
      text-transform: uppercase; letter-spacing: 0.08em;
      color: #9898a8; margin-bottom: 0.4rem;
    }
    .cv-tags { display: flex; flex-wrap: wrap; gap: 5px; }
    .cv-tag {
      font-family: 'DM Mono', monospace; font-size: 0.7rem;
      background: #ededf5; color: #2a2a6e;
      padding: 3px 8px; border-radius: 3px;
    }
    .cv-tag.hot {
      background: #e8edff; color: #1a3cff;
      border: 1px solid rgba(26,60,255,0.2);
    }
    /* Education */
    .cv-edu { margin-bottom: 1rem; }
    .cv-edu-deg { font-size: 0.87rem; font-weight: 600; color: #0e0e10; }
    .cv-edu-school { font-size: 0.79rem; color: #1a3cff; margin-top: 1px; }
    .cv-edu-year {
      font-family: 'DM Mono', monospace; font-size: 0.7rem;
      color: #9898a8; margin-top: 1px;
    }
    /* Languages */
    .cv-lang {
      display: flex; justify-content: space-between;
      align-items: center; margin-bottom: 0.6rem;
    }
    .cv-lang-name { font-size: 0.83rem; color: #0e0e10; }
    .cv-lang-lvl {
      font-family: 'DM Mono', monospace; font-size: 0.66rem;
      color: #9898a8; letter-spacing: 0.05em;
    }
    .cv-dots { display: flex; gap: 3px; margin-top: 3px; }
    .cv-dot { width: 7px; height: 7px; border-radius: 50%; background: #e2e0da; }
    .cv-dot.on { background: #1a3cff; }
    """

    # ── Helper builders ───────────────────────────────────────────────
    def build_dots(score, max_score=5):
        return ''.join(
            f'<div class="cv-dot{\" on\" if i < score else \"\"}"></div>'
            for i in range(max_score)
        )

    def build_tags(items, hot=False):
        cls = 'cv-tag hot' if hot else 'cv-tag'
        return ''.join(f'<span class="{cls}">{t}</span>' for t in items)

    def build_job(j):
        achievements = ''.join(f'<li>{r}</li>' for r in j['realisations'])
        return f"""
        <article class="cv-job">
          <div class="cv-job-head">
            <span class="cv-job-title">{j['titre']}</span>
            <span class="cv-job-period">{j['periode']}</span>
          </div>
          <div class="cv-job-company">{j['entreprise']}</div>
          <p class="cv-job-desc">{j['description']}</p>
          <ul class="cv-achievements">{achievements}</ul>
        </article>"""

    def build_project(p):
        return f"""
        <article class="cv-job">
          <div class="cv-job-head">
            <span class="cv-job-title">{p['nom']}</span>
            <span class="cv-job-period">{p['contexte']}</span>
          </div>
          <div class="cv-job-company">{p['lien']}</div>
          <p class="cv-job-desc">{p['description']}</p>
        </article>"""

    # ── Assemble HTML ────────────────────────────────────────────────
    jobs_html    = ''.join(build_job(j) for j in experiences)
    projs_html   = ''.join(build_project(p) for p in projets)

    stats_html   = ''.join(
        f'<div class="cv-stat"><span class="cv-stat-num">{s["valeur"]}</span>'
        f'<span class="cv-stat-lbl">{s["label"]}</span></div>'
        for s in stats
    )

    skills_html  = ''.join(
        f'<div class="cv-tag-group">'
        f'<div class="cv-tag-group-lbl">{cat}</div>'
        f'<div class="cv-tags">{build_tags(v["hot"], hot=True)}{build_tags(v["other"])}</div>'
        f'</div>'
        for cat, v in competences.items()
    )

    edu_html     = ''.join(
        f'<div class="cv-edu">'
        f'<div class="cv-edu-deg">{f["diplome"]}</div>'
        f'<div class="cv-edu-school">{f["ecole"]}</div>'
        f'<div class="cv-edu-year">{f["annee"]}</div>'
        f'</div>'
        for f in formations
    )

    lang_html    = ''.join(
        f'<div class="cv-lang">'
        f'<div><div class="cv-lang-name">{l["langue"]}</div>'
        f'<div class="cv-dots">{build_dots(l["score"])}</div></div>'
        f'<div class="cv-lang-lvl">{l["niveau"]}</div>'
        f'</div>'
        for l in langues
    )

    interests_html = build_tags(interets)

    # ── Full HTML ────────────────────────────────────────────────────
    p = profil
    html = f"""
    <style>{css}</style>
    <div class="cv-wrap">
      <div class="cv-page">

        <!-- HEADER -->
        <header class="cv-header">
          <div>
            <h1 class="cv-name">{p['nom'].split()[0]} <em>{p['nom'].split()[1]}</em></h1>
            <p class="cv-role">// {p['titre']}</p>
            <p class="cv-summary">{p['resume']}</p>
            <span class="cv-badge">{p['annees_xp']} ANS D'EXPÉRIENCE</span>
          </div>
          <div class="cv-contact">
            <div>{p['email']}</div>
            <div>{p['telephone']}</div>
            <div>{p['ville']}</div>
            <div><a href="#">{p['linkedin']}</a></div>
            <div><a href="#">{p['github']}</a></div>
          </div>
        </header>

        <div class="cv-body">

          <!-- MAIN -->
          <main class="cv-main">
            <section class="cv-section">
              <h2 class="cv-section-title">Expérience professionnelle</h2>
              {jobs_html}
            </section>
            <section class="cv-section">
              <h2 class="cv-section-title">Projets &amp; Open Source</h2>
              {projs_html}
            </section>
          </main>

          <!-- ASIDE -->
          <aside class="cv-aside">
            <section class="cv-section">
              <h2 class="cv-section-title">En chiffres</h2>
              <div class="cv-stats">{stats_html}</div>
            </section>
            <section class="cv-section">
              <h2 class="cv-section-title">Compétences</h2>
              {skills_html}
            </section>
            <section class="cv-section">
              <h2 class="cv-section-title">Formation</h2>
              {edu_html}
            </section>
            <section class="cv-section">
              <h2 class="cv-section-title">Langues</h2>
              {lang_html}
            </section>
            <section class="cv-section">
              <h2 class="cv-section-title">Intérêts</h2>
              <div class="cv-tags">{interests_html}</div>
            </section>
          </aside>

        </div>
      </div>
    </div>
    """
    return HTML(html)

display(render_cv())
```

## 💾 3. Export — Sauvegarder le CV en fichier HTML

```python
from IPython.display import FileLink

# Récupérer le HTML brut (sans balise <style> dupliquée)
def export_cv_html(filename="cv_amadou_diallo.html"):
    html_obj = render_cv()
    full_html = f"""<!DOCTYPE html>
<html lang='fr'>
<head>
  <meta charset='UTF-8'>
  <meta name='viewport' content='width=device-width, initial-scale=1.0'>
  <title>CV — {profil['nom']}</title>
</head>
<body style='margin:0;background:#f8f7f4'>
{html_obj.data}
</body>
</html>"""

    with open(filename, "w", encoding="utf-8") as f:
        f.write(full_html)

    print(f"✅ CV exporté : {filename}")
    return FileLink(filename, result_html_prefix="📥 Télécharger : ")

export_cv_html()
```

```python
# ── Timeline de carrière ──────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 3.5))
fig.patch.set_facecolor('#f8f7f4')
ax.set_facecolor('#f8f7f4')

timeline = [
    (2013, 2017, "Gainde 2000",          "Dev Full-Stack"),
    (2017, 2020, "Wave Mobile Money",     "Senior Backend"),
    (2020, 2026, "Orange Digital Center", "Lead Engineer"),
]

colors = ['#ededf5', '#e8edff', '#1a3cff']
text_colors = ['#2a2a6e', '#1a3cff', '#ffffff']

for i, (start, end, company, role) in enumerate(timeline):
    ax.barh(0, end - start, left=start, height=0.55,
            color=colors[i], edgecolor='#1a3cff', linewidth=1)
    mid = start + (end - start) / 2
    ax.text(mid, 0.05, company, ha='center', va='center',
            fontsize=9, fontweight='700', color=text_colors[i])
    ax.text(mid, -0.18, role, ha='center', va='center',
            fontsize=7.5, color=text_colors[i], style='italic')

ax.set_xlim(2012, 2027)
ax.set_ylim(-0.6, 0.6)
ax.set_yticks([])
ax.set_xticks(range(2013, 2027, 1))
ax.set_xticklabels([str(y) for y in range(2013, 2027, 1)],
                   fontsize=8.5, color='#5a5a6a')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_color('#e2e0da')
ax.tick_params(axis='x', colors='#9898a8', length=4)

ax.set_title("Timeline de carrière", fontsize=12, fontweight='700',
             color='#0e0e10', pad=12, loc='left')

plt.tight_layout()
plt.savefig('timeline_carriere.png', dpi=150, bbox_inches='tight', facecolor='#f8f7f4')
plt.show()
print("✅ Timeline sauvegardée : timeline_carriere.png")
```
